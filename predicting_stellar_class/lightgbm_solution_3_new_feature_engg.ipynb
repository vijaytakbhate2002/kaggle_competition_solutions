```python
ID = 'id'
TARGET = 'class'
train_df[TARGET] = train_df[TARGET].map({'GALAXY': 0, 'QSO': 1, 'STAR': 2})
X = train_df.drop([ID, TARGET], axis=1); train_id = train_df[ID]
y = train_df[TARGET]
X_test = test_df.drop([ID], axis=1); test_id = test_df[ID]
del train_df, test_df
print("X      init shape:", X.shape)
print("X_test init shape:", X_test.shape, "\n")

cat_cols = X.select_dtypes(include=['object']).columns.tolist()
num_cols = X.select_dtypes(exclude=['object']).columns.tolist()
print("init len(cat_cols):", len(cat_cols))
print("init len(num_cols):", len(num_cols), "\n")

category_map = {}
color_pairs = [
    ('u', 'g'),
    ('g', 'r'),
    ('g', 'i'),
    ('r', 'i'),
    ('i', 'z'),
    ('u', 'r'),
    ('g', 'z'),
    ('u', 'z'),
]
important_combos = [
    ('alpha_cat_', 'delta_cat_'),
    ('u_cat_', 'z_cat_'),
]
important_combos = sorted(important_combos)
def feature_engineering(df, fit=False):
    # Arithmetic interaction
    df['_g_/_redshift'] = (df['g'] / (df['redshift'] + 1e-6)).astype('float32')
    df['_i_/_redshift'] = (df['i'] / (df['redshift'] + 1e-6)).astype('float32')
    
    # ── Color indices ──
    for a, b in color_pairs:
        df[f"_{a}-{b}"] = (df[a] - df[b]).astype('float32')

    # ── Redshift-based features ──
    df['_log_redshift']   = np.log1p(np.abs(df['redshift'])).astype('float32')
    df['_redshift_sq']    = (df['redshift'] ** 2).astype('float32')
    df['_redshift_sign']  = np.sign(df['redshift']).astype('float32')
    df['_is_low_z']       = (np.abs(df['redshift']) < 0.002).astype('float32')
    df['_is_high_z']      = (df['redshift'] > 0.3).astype('float32')

    # ── Magnitude statistics ──
    mag_cols = ['u', 'g', 'r', 'i', 'z']
    df['_mag_mean']   = df[mag_cols].mean(axis=1).astype('float32')
    df['_mag_std']    = df[mag_cols].std(axis=1).astype('float32')
    df['_mag_range']  = (df[mag_cols].max(axis=1) - df[mag_cols].min(axis=1)).astype('float32')
    df['_mag_skew']   = df[mag_cols].skew(axis=1).astype('float32')

    # ── Sky position features ──
    alpha_rad = np.radians(df['alpha'])
    delta_rad = np.radians(df['delta'])
    df['_sin_alpha'] = np.sin(alpha_rad).astype('float32')
    df['_cos_alpha'] = np.cos(alpha_rad).astype('float32')
    df['_sin_delta'] = np.sin(delta_rad).astype('float32')
    df['_cos_delta'] = np.cos(delta_rad).astype('float32')

    # ── Galactic latitude proxy ──
    df['_gal_lat_proxy'] = np.abs(df['delta'] - 27.1).astype('float32')

    # ── Interactions with redshift ──
    df['_redshift_x_u_g']  = (df['redshift'] * df['_u-g']).astype('float32')
    df['_redshift_x_g_r']  = (df['redshift'] * df['_g-r']).astype('float32')
    df['_redshift_x_r_i']  = (df['redshift'] * df['_r-i']).astype('float32')

    # ── Features from second-look-treeees ──
    df['_uv_vs_old_diff'] = (df['_u-r'] - df['_g-r']).astype('float32')
    df['_uv_vs_old_ratio'] = (df['_u-r'] / (df['_g-r'] + 1e-5)).astype('float32')
    df['_g-r_squared'] = (df['_g-r'] ** 2).astype('float32')
    df['_u-r_squared'] = (df['_u-r'] ** 2).astype('float32')
    df['_r_times_g-r'] = (df['r'] * df['_g-r']).astype('float32')
    df['_r_times_u-r'] = (df['r'] * df['_u-r']).astype('float32')
    df['_total_magnitude_proxy'] = (df['u'] + df['g'] + df['r'] + df['i'] + df['z']).astype('float32')

    # ── Encode categorical features (XGBoost manual style) ──
    spe_map = {'M':0, 'O/B':1, 'G/K':2, 'A/F':3}
    gal_map = {'Red_Sequence':0, 'Blue_Cloud':1}
    if 'spectral_type' in df.columns:
        df['spe_'] = df['spectral_type'].map(spe_map).fillna(-1).astype('int32')
        df['spe_'] = df['spe_'].astype('category')
    if 'galaxy_population' in df.columns:
        df['gal_'] = df['galaxy_population'].map(gal_map).fillna(-1).astype('int32')
        df['gal_'] = df['gal_'].astype('category')

    # ── Spectral type interactions ──
    if 'spectral_type' in df.columns:
        hot_star  = df['spectral_type'].isin(['O/B', 'A/F']).astype('float32')
        cool_star = (df['spectral_type'] == 'M').astype('float32')
        df['_hot_star'] = hot_star
        df['_cool_star'] = cool_star
        df['_hot_x_lowz'] = (hot_star * df['_is_low_z']).astype('float32')

    # Categorize string cats
    for col in cat_cols:
        if fit:
            codes, uniques = df[col].factorize()
            category_map[col] = uniques
        else:
            uniques = category_map[col]
            code_map = {cat: i for i, cat in enumerate(uniques)}
            codes = df[col].map(code_map).fillna(-1).astype('int32')
        df[col] = codes
        df[col] = df[col].astype('category')

    # Categorize numericals
    for col in num_cols:
        cat_name = f"{col}_cat_"
        if fit:
            codes, uniques = np.floor(df[col]).factorize()
            category_map[col] = uniques
        else:
            uniques = category_map[col]
            code_map = {cat: i for i, cat in enumerate(uniques)}
            codes = np.floor(df[col]).map(code_map).fillna(-1).astype('int32')
        df[cat_name] = codes
        df[cat_name] = df[cat_name].astype('category')

    # Discretize numericals
    bin_config = {'delta': [100, 500]}
    for col, bins_list in bin_config.items():
        for n_bins in bins_list:
            for strategy in ['quantile']:
                bin_name = f"{col}_{n_bins}_{strategy}_bin_"
                if fit:
                    kb = KBinsDiscretizer(
                        n_bins=n_bins,
                        encode='ordinal',
                        strategy=strategy,
                        subsample=None
                    )
                    binned = kb.fit_transform(df[[col]]).ravel().astype('int32')
                    category_map[bin_name] = kb
                else:
                    kb = category_map[bin_name]
                    binned = kb.transform(df[[col]]).ravel().astype('int32')
                df[bin_name] = binned
                df[bin_name] = df[bin_name].astype('category')

    # Create interaction categories
    combo_names = []
    for cols in important_combos:
        combo_name = '_'.join(cols) + '_'
        combo_names.append(combo_name)
        combo_series = df[cols[0]].astype(str)
        for col in cols[1:]:
            combo_series = combo_series + '_' + df[col].astype(str)
        if fit:
            codes, uniques = pd.factorize(combo_series, sort=False)
            category_map[combo_name] = uniques
        else:
            uniques = category_map[combo_name]
            code_map = {cat: i for i, cat in enumerate(uniques)}
            codes = combo_series.map(code_map).fillna(-1).astype('int32')
        df[combo_name] = codes
        df[combo_name] = df[combo_name].astype('category')

    new_cat_cols = [col for col in df.columns if col.endswith('_') and col not in cat_cols and col not in num_cols]
    new_num_cols = [col for col in df.columns if col.startswith('_') and col not in cat_cols and col not in num_cols]
    return df, new_cat_cols, new_num_cols, combo_names

X, new_cat_cols, new_num_cols, combo_names = feature_engineering(X, fit=True)
X_test, new_cat_cols, new_num_cols, combo_names = feature_engineering(X_test, fit=False)
cat_cols += new_cat_cols; num_cols += new_num_cols
print("len(new_cat_cols):", len(new_cat_cols))
print("len(new_num_cols):", len(new_num_cols), "\n")

cat_cols = sorted(cat_cols)
X = X.reindex(sorted(X.columns), axis=1)
X_test = X_test.reindex(sorted(X_test.columns), axis=1)
print("prep len(cat_cols):", len(cat_cols))
print("prep len(num_cols):", len(num_cols), "\n")
print("X      prep shape:", X.shape)
print("X_test prep shape:", X_test.shape, "\n")
```

## Binary Chain 
- Train First LightGBM (Galaxy Detection)
- Train Second LightGBM (QSQ Detection)

```python
gal_det_best_params = {'n_estimators': 1855, 'learning_rate': 0.024262368681948563, 'num_leaves': 113, 'max_depth': 16, 
                       'min_child_samples': 12, 'subsample': 0.8584322598224993, 'colsample_bytree': 0.6424451606563617, 
                       'reg_alpha': 2.7787822410602394, 'reg_lambda': 0.007288115036802227, 'min_split_gain': 0.4024979374524029}

qso_det_best_params = {'n_estimators': 2000, 'learning_rate': 0.03571369122147518, 'num_leaves': 47, 'max_depth': 13, 
                       'min_child_samples': 16, 'subsample': 0.9052622034023979, 'colsample_bytree': 0.6218958111549964, 
                       'reg_alpha': 8.163535304194495, 'reg_lambda': 0.007338868506875836, 'min_split_gain': 0.9521234850775032}


class Chain:
    def __init__(self, gal_params, qsq_params):
        self.gal_params = gal_params
        self.qsq_params = qsq_params
        self.gal_det_model = None
        self.qsq_det_model = None

    def fit(self, X, y):
        gal_det_y = np.where(y == TARGET_MAPPING['GALAXY'], 1, 0)
        qsq_det_y = np.where(y == TARGET_MAPPING['QSO'], 1, 0)

        self.gal_params['class_weight'] = 'balanced'
        self.gal_params['n_estimators'] = 10000
        self.qsq_params['class_weight'] = 'balanced'
        self.qsq_params['n_estimators'] = 10000

        self.gal_det_model = lgb.LGBMClassifier(**self.gal_params)
        self.qsq_det_model = lgb.LGBMClassifier(**self.qsq_params)

        self.gal_det_model.fit(X, gal_det_y)
        self.qsq_det_model.fit(X, qsq_det_y)
        
        return self

    def galaxy_detector(self, X):
        gal_y_pred = self.gal_det_model.predict(X)
        gal_y_prob = self.gal_det_model.predict_proba(X)[:,1]

        return gal_y_pred, gal_y_prob
    
    def qsq_detector(self, X):
        qsq_y_pred = self.qsq_det_model.predict(X)
        qsq_y_prob = self.qsq_det_model.predict_proba(X)[:,1]

        return qsq_y_pred, qsq_y_prob
    
    def predict(self, X):

        gal_pred, _ = self.galaxy_detector(X)
        qsq_pred, _ = self.qsq_detector(X)

        y_pred = np.full(len(X), TARGET_MAPPING['STAR'], dtype=int)
        y_pred[gal_pred == 1] = TARGET_MAPPING['GALAXY']

        qsq_mask = (gal_pred == 0) & (qsq_pred == 1)
        y_pred[qsq_mask] = TARGET_MAPPING['QSO']

        return y_pred

    def predict_proba(self, X):

        gal_prob = self.gal_det_model.predict_proba(X)[:, 1]
        qsq_prob = self.qsq_det_model.predict_proba(X)[:, 1]

        star_prob = (1 - gal_prob) * (1 - qsq_prob)

        probs = np.column_stack([gal_prob, qsq_prob, star_prob])
        probs = probs / probs.sum(axis=1, keepdims=True)

        return probs

# lgb_chain = Chain(params_lg2, params_lg2)
# lgb_chain.fit(X_train, y_train)
# y_pred = lgb_chain.predict(X_test)
# balanced_acc = balanced_accuracy_score(y_test, y_pred)
# print(f"Chain Score {balanced_acc:>20.4f}")

# lgb_chain = Chain(params_lg2, params_lg2)
# lgb_chain.fit(X, y)
```
