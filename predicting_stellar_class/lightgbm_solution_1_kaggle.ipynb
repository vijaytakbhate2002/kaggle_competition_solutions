| Dataset | GALAXY | QSO | STAR | Total |

| **Extra** | 59.45% | 18.96% | 21.59% | 100,000 |

**<span style="color: #2E86AB;">🔍 Key Observations:</span>**

1. **Class Imbalance**: GALAXY is the dominant class (65% in training data), while STAR is the minority (14%)
2. **Distribution Shift**: The external data has a different distribution, particularly more STAR samples
3. **Sample Size**: With 577K training samples, we have sufficient data for robust model training

**Numerical Features (8):**
- `alpha`, `delta`: Celestial coordinates
- `u`, `g`, `r`, `i`, `z`: Magnitude measurements in 5 wavelength bands
- `redshift`: Redshift value indicating distance/velocity

**Categorical Features (2):**
- `spectral_type`: Stellar spectral classification (M, G/K, A/F, O/B)
- `galaxy_population`: Galaxy classification (Blue_Cloud, Red_Sequence)

**<span style="color: #2E86AB;">💡 Strategic Insight:</span>** The class imbalance suggests we need to use:
- **Balanced class weights** in LightGBM configuration
- **Stratified cross-validation** to maintain distribution consistency
- **Balanced accuracy** as the evaluation metric (rather than simple accuracy)

# ⚙️ <span style="color: #2E86AB;">Advanced Feature Engineering</span>

<div style="background: linear-gradient(135deg, #fc4a1a 0%, #f7b733 100%); padding: 20px; border-radius: 10px; margin: 20px 0;">
    <h3 style="color: white; margin: 0;">🚀 Transforming Raw Stellar Data into Predictive Features</h3>
</div>

## 🎯 Feature Engineering Strategy Overview

This section implements a sophisticated feature engineering pipeline that transforms raw astronomical measurements into powerful predictive features for the LightGBM model:

### <span style="color: #E74C3C;">Feature Expansion Results:</span>

| Dataset | Before FE | After FE | New Features Added |
|---------|-----------|----------|-------------------|
| **Train** | 12 columns | 61 columns | **+49 features** |
| **Test** | 11 columns | 60 columns | **+49 features** |

**<span style="color: #2E86AB;">🎨 Feature Categories:</span>**

**<span style="background: #F8F9FA; padding: 5px 10px; border-radius: 5px;">🔥 Astrophysical Ranking Features</span>**
- `spectral_rank_fe`: Ordinal encoding of spectral types (M=1, G/K=2, A/F=3, O/B=4)
- `population_rank_fe`: Galaxy population ranking (Red_Sequence=0, Blue_Cloud=1)
- `astro_score_fe`: Composite score combining spectral and population characteristics

**<span style="background: #F8F9FA; padding: 5px 10px; border-radius: 5px;">🎯 Candidate Identification Features</span>**
- `is_red_galaxy_candidate_fe`: Binary indicator for likely red galaxy candidates
- `is_qso_candidate_fe`: Binary indicator for likely QSO (quasar) candidates
- These features capture domain knowledge from the categorical analysis

**<span style="background: #F8F9FA; padding: 5px 10px; border-radius: 5px;">🌈 Photometric Filter Features</span>**
- `dom_fltr_fe`: Dominant wavelength filter (minimum magnitude = brightest)
- Filter statistics: `mean_fltr_fe`, `max_fltr_fe`, `min_fltr_fe`, `std_fltr_fe`, `med_fltr_fe`
- Capture overall brightness patterns across all 5 wavelength bands

**<span style="background: #F8F9FA; padding: 5px 10px; border-radius: 5px;">🎨 Color Index Features</span>**
- All pairwise differences: `u_g`, `u_r`, `u_i`, `u_z`, `g_r`, `g_i`, `g_z`, `r_i`, `r_z`, `i_z`
- Color curvature features: `ug_gr`, `ug_ri`, `ug_iz`, `gr_ri`, `gr_iz`, `ri_iz`
- These capture spectral energy distribution characteristics

**<span style="background: #F8F9FA; padding: 5px 10px; border-radius: 5px;">🔭 Redshift Interaction Features</span>**
- Distance-scaled color indices: `u_g_redshift`, `g_r_redshift`, etc.
- Capture how spectral characteristics change with distance
- Important for cosmological effects on stellar classification

**<span style="background: #F8F9FA; padding: 5px 10px; border-radius: 5px;">📐 Coordinate Transformation Features</span>**
- **Cartesian coordinates**: `coord_x_fe`, `coord_y_fe`, `coord_z_fe` from equatorial coordinates
- **Distance-scaled coordinates**: `coord_x_redshift_fe`, etc.
- **Galactic coordinates**: `gal_x_fe`, `gal_y_fe`, `gal_z_fe` using J2000 transformation
- Enable spatial analysis in different coordinate systems

**<span style="background: #F8F9FA; padding: 5px 10px; border-radius: 5px;">🔬 Specialized Astronomical Features</span>**
- `stellar_color_bin_fe`: Discretized stellar color temperature bins
- `dom_redshift_magn_fe`: Dominant filter magnitude scaled by redshift
- Interaction between brightest wavelength and distance

**<span style="background: #F8F9FA; padding: 5px 10px; border-radius: 5px;">🔗 Categorical Interaction Features</span>**
- Combined categorical feature: `spectral_type_fe_&_galaxy_population_fe`
- Captures interaction between stellar type and galaxy environment

### <span style="color: #E74C3C;">Encoding Strategy:</span>

The function implements **Target Encoding** for categorical features:
- **OrdinalEncoder**: Converts categorical features to numerical values
- **Training mode**: Fits encoder on training data and transforms
- **Inference mode**: Uses pre-fitted encoder for test data
- **Target mapping**: Converts string classes to numerical labels (GALAXY=0, QSO=1, STAR=2)

### <span style="color: #2E86AB;">💡 Why This Feature Engineering Works:</span>

1. **Domain Knowledge Integration**: Features incorporate astronomical understanding
2. **Non-linear Relationships**: Color indices and interactions capture complex patterns
3. **Scale Invariance**: Statistical features work across different magnitude ranges

## 📊 Target Encoding Implementation

This section implements **K-Fold Target Encoding** to transform categorical features into numerical representations while preventing data leakage:

### <span style="color: #E74C3C;">Target Encoding Purpose:</span>

**<span style="background: #F8F9FA; padding: 5px 10px; border-radius: 5px;">🎯 Converting Categories to Numbers</span>**
- Replaces categorical values with the mean target value for that category
- Captures the relationship between categorical features and the target variable
- More informative than simple one-hot encoding for high-cardinality features

**<span style="background: #F8F9FA; padding: 5px 10px; border-radius: 5px;">🛡️ Preventing Data Leakage</span>**
- Uses **K-Fold strategy** to calculate encoding statistics
- Each fold's encoding uses data from other folds only
- Ensures validation/test data doesn't influence training encoding

### <span style="color: #E74C3C;">Encoding Process:</span>

**<span style="background: #F8F9FA; padding: 5px 10px; border-radius: 5px;">🔄 Training Data Encoding</span>**
1. Split training data into K folds (K=5 in this implementation)
2. For each fold:
   - Calculate target mean for each categorical value using **other folds**
   - Apply these statistics to the current fold
   - Repeat for all categorical features and target classes
3. Combine all folds to get fully encoded training data

**<span style="background: #F8F9FA; padding: 5px 10px; border-radius: 5px;">📈 Test Data Encoding</span>**
1. Use statistics calculated from the **entire training set**
2. Apply same encoding mappings to test data
3. Ensures consistent encoding between train and test

### <span style="color: #E74C3C;">Feature Expansion Results:</span>

| Dataset | Before Encoding | After Encoding | New Features |
|---------|----------------|----------------|--------------|

  - `spectral_type_fe_GALAXY`, `spectral_type_fe_QSO`, `spectral_type_fe_STAR`
