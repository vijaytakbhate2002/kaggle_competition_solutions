```python
def merge_dataframes_in_folder(folder_path, id_col='id'):
    file_patterns = ['*.csv', '*.xlsx', '*.xls']
    dfs = []
    for pattern in file_patterns:
        for filepath in glob.glob(os.path.join(folder_path, pattern)):
            try:
                if filepath.lower().endswith('.csv'):
                    df = pd.read_csv(filepath)
                else:
                    df = pd.read_excel(filepath)
            except Exception as exc:
                print(f"Skipping {filepath}: {exc}")
                continue

            if id_col not in df.columns:
                print(f"Skipping {filepath}: missing column {id_col}")
                continue

            prefix = os.path.splitext(os.path.basename(filepath))[0]
            df = df.rename(columns={c: f"{prefix}_{c}" for c in df.columns if c != id_col})
            dfs.append(df)

    if not dfs:
        raise ValueError("No valid dataframes found in the folder")

    merged_df = dfs[0]
    for df in dfs[1:]:
        merged_df = merged_df.merge(df, on=id_col, how='outer')

    return merged_df

final_blend = merge_dataframes_in_folder("submissions")
final_blend
```

```python
final_blend_avg = final_blend.drop(columns=['id']).mean(axis=1)
id_col = final_blend['id']

final_blend_avg = pd.DataFrame({'id': id_col, 'target': final_blend_avg})
final_blend_avg.to_csv('final_blend_avg.csv', index=False)
final_blend_avg
```
